```python
# ============================================
# HARD & SOFT CLUSTERING
# K-Means + Gaussian Mixture Model (GMM)
# ============================================

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture

# --------------------------------------------
# 1. Create sample dataset
# --------------------------------------------
X, _ = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.2,
    random_state=42
)

# --------------------------------------------
# 2. HARD CLUSTERING - K MEANS
# --------------------------------------------
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)

# Predict clusters for existing data
hard_clusters = kmeans.predict(X)

print("========== HARD CLUSTERING ==========")
print("First 10 cluster predictions:")
print(hard_clusters[:10])


# --------------------------------------------
# 3. Predict cluster for NEW DATA
# --------------------------------------------
new_point = [[5, 5]]

hard_prediction = kmeans.predict(new_point)

print("\nNew Data Point:", new_point)
print("Hard Cluster Prediction:", hard_prediction[0])


# --------------------------------------------
# 4. SOFT CLUSTERING - GMM
# --------------------------------------------
gmm = GaussianMixture(
    n_components=3,
    random_state=42
)

gmm.fit(X)

# Calculate probability of belonging
# to every cluster
probabilities = gmm.predict_proba(new_point)

print("\n========== SOFT CLUSTERING ==========")
print("New Data Point:", new_point)

for i, probability in enumerate(probabilities[0]):
    print(
        f"Cluster {i}: "
        f"{probability * 100:.2f}%"
    )


# --------------------------------------------
# 5. Convert soft prediction to hard prediction
# --------------------------------------------
soft_hard_prediction = np.argmax(probabilities[0])

print(
    "\nMost Likely Cluster:",
    soft_hard_prediction
)
```
